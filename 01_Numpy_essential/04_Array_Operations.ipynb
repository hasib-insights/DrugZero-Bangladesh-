## Arithmetic Operations
Perform element-wise mathematical operations between arrays for fast numerical computation.

````python
import numpy as np
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])
````

````python
# Addition (+)
print(a + b)
````

````python
# Subtraction (-)
print(a - b)
````

````python
# Multiplication (*) - element-wise, NOT matrix multiplication
print(a * b)
````

````python
# Division (/)
print(b / a)
print(a / b)
````

````python
# Floor Division (//) - integer division
print(b // a)
````

````python
# Power (**)
print(a ** 2)
print(a ** b)
````

````python
# Modulo (%)
print(b % a)
````

```
   5 % 1:
     5 ÷ 1 = 5  (exactly, no remainder)
     → 0

   6 % 2:
     6 ÷ 2 = 3  (exactly, no remainder)
     → 0

   7 % 3:
     7 ÷ 3 = 2  remainder 1  (2×3=6, 7−6=1)
     → 1

   8 % 4:
     8 ÷ 4 = 2  (exactly, no remainder)
     → 0
     

````python
# Scalar operations (broadcasting)
print(a + 10)
print(a * 2)
print(a ** 2)
````

>Key Concept: All operations are element-wise by default. For matrix multiplication, use `np.dot()` or `@` operator.

## Comparison Operations

````python
a = np.array([1, 2, 3, 4, 5])

print(a > 3)
````

````python
print(a == 3)
````

````python
print(a != 3)
````

````python
# Boolean indexing
print(a[a > 3])
````

````python
# Boolean indexing
print(a[(a > 2) & (a < 5)])
````

## Aggregate Functions

````python
arr = np.array([1, 2, 3, 4, 5])

print('sum:', arr.sum())
print('mean:',arr.mean())
print('std:',arr.std())
print('var:',arr.var())
print('min:',arr.min())
print('max:',arr.max())
print('argmin:',arr.argmin())
print('argmin:',arr.argmax())
print('size:', len(arr))
````

# Master Table

| # | Function | Output | Formula / Calculation | Meaning |
|:-:|----------|:------:|----------------------|---------|
| 1 | `arr.sum()` | **15** | `1 + 2 + 3 + 4 + 5 = 15` | Total of all elements |
| 2 | `arr.mean()` | **3.0** | `15 / 5 = 3.0` | Average |
| 3 | `arr.var()` | **2.0** | `(4 + 1 + 0 + 1 + 4) / 5 = 10 / 5 = 2.0` | Variance |
| 4 | `arr.std()` | **1.414** | `√var = √2 ≈ 1.414` | Standard deviation |
| 5 | `arr.min()` | **1** | Smallest value | Minimum value |
| 6 | `arr.max()` | **5** | Largest value | Maximum value |
| 7 | `arr.argmin()` | **0** | Index where value = 1 | Index of smallest |
| 8 | `arr.argmax()` | **4** | Index where value = 5 | Index of largest |
| 9 | `arr.size` | **5** | `len(arr) = 5` | Total element count |


````python
# For 2D arrays
arr2d = np.array([[1, 2, 3], [4, 5, 6]])

print(arr2d.sum(axis=0))  # Output: [5 7 9] (sum along columns)
print(arr2d.sum(axis=1))  # Output: [6 15] (sum along rows)
````

```
   axis=0 (DOWN ↓)                    axis=1 (ACROSS →)

   ├───┼───┼───┤                      ├───┼───┼───┤
   │ 4 │ 5 │ 6 │                      │ 4 │ 5 │ 6 │ → 15
   └───┴───┴───┘                      └───┴───┴───┘
     ↓   ↓   ↓
     5   7   9                        Result: [6, 15]
                                      shape: (2,)
